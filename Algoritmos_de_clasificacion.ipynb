División entrenamiento / prueba

```python

from sklearn.datasets import load_breast_cancer
import pandas as pd

# Cargar el dataset en formato DataFrame
data = load_breast_cancer(as_frame=True)

# Obtener las variables predictoras
X = data.data.copy()

# Convertir la clase positiva en 'maligno'
y = (data.target == 0).astype(int)

# Mostrar las dimensiones y la distribución de las clases
print("Dimensiones de X:", X.shape)
print("\nDistribución de las clases:")
print(y.value_counts())

```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train.shape, X_test.shape)
```

Función común de evaluación (función para evitar repetir código).

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

def evaluar_modelo(nombre, modelo, X_eval, y_eval):
    pred = modelo.predict(X_eval)

    resultados = {
        'Modelo': nombre,
        'Accuracy': accuracy_score(y_eval, pred),
        'Precision': precision_score(y_eval, pred, zero_division=0),
        'Recall': recall_score(y_eval, pred, zero_division=0),
        'F1': f1_score(y_eval, pred, zero_division=0)
    }

    print("Modelo:", nombre)
    print("\nMatriz de confusión:")
    print(confusion_matrix(y_eval, pred))

    print("\nMétricas de evaluación:")
    for metrica, valor in resultados.items():
        if metrica != 'Modelo':
            print(f"{metrica}: {valor:.4f}")

    return resultados
```

Algoritmo Decision Tree:

```python

from sklearn.tree import DecisionTreeClassifier

# Crear el modelo
modelo_dt = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

# Entrenar el modelo con los datos de entrenamiento
modelo_dt.fit(X_train, y_train)

# Evaluar el modelo con los datos de prueba
resultados_dt = evaluar_modelo(
    "Decision Tree",
    modelo_dt,
    X_test,
    y_test
)
```

**Interpretación de resultados:** El modelo Decision Tree obtuvo un Accuracy del 91,23 %, lo que indica que clasificó correctamente la mayoría de las muestras. La Precision fue del 94,44 %, mientras que el Recall alcanzó el 80,95 %, identificando correctamente 34 de los 42 tumores malignos. El F1-score fue del 87,18 %, reflejando el equilibrio entre Precision y Recall. La matriz de confusión muestra 8 falsos negativos, es decir, tumores malignos clasificados erróneamente como benignos, y 2 falsos positivos. Estos resultados indican un desempeño aceptable, aunque existe margen de mejora en la detección de tumores malignos.

Algoritmo Random Forest:

```python

from sklearn.ensemble import RandomForestClassifier

# Crear el modelo con 100 árboles
modelo_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Entrenar el modelo con los mismos datos de entrenamiento
modelo_rf.fit(X_train, y_train)

# Evaluar el modelo con los mismos datos de prueba
resultados_rf = evaluar_modelo(
    "Random Forest",
    modelo_rf,
    X_test,
    y_test
)
```

**Interpretación de resultados:** El modelo Random Forest obtuvo un Accuracy del 97,37 %, una Precision del 100 %, un Recall del 92,86 % y un F1-score del 96,30 %. Estos resultados indican un excelente desempeño en la clasificación de las muestras. La matriz de confusión muestra 39 tumores malignos identificados correctamente, 72 tumores benignos clasificados correctamente y solo 3 falsos negativos, sin falsos positivos. Por tanto, el modelo logró identificar la mayoría de los tumores malignos, aunque no detectó todos los casos.

Algoritmo Support Vector Machine - SVM:

```python

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

modelo_svm = make_pipeline(
    StandardScaler(),
    SVC(kernel='rbf', C=1.0, gamma='scale')
)

# Entrenar el modelo
modelo_svm.fit(X_train, y_train)

# Evaluar el modelo
resultados_svm = evaluar_modelo(
    "Support Vector Machine (SVM)",
    modelo_svm,
    X_test,
    y_test
)
```

**Interpretación de resultados:** El modelo SVM obtuvo un Accuracy del 97,37 %, una Precision del 100 %, un Recall del 92,86 % y un F1-score del 96,30 %, mostrando un excelente desempeño. La matriz de confusión indica que clasificó correctamente 72 tumores benignos y 39 malignos, sin falsos positivos. Sin embargo, presentó 3 falsos negativos, es decir, tres tumores malignos fueron clasificados erróneamente como benignos. En general, el modelo logró una alta precisión y una buena capacidad para identificar tumores malignos, aunque no detectó todos los casos.

Algoritmo K-Nearest Neighbors-K-NN:

```python

from sklearn.neighbors import KNeighborsClassifier

modelo_knn = make_pipeline(
    StandardScaler(),
    KNeighborsClassifier(n_neighbors=5)
)

# Entrenar el modelo
modelo_knn.fit(X_train, y_train)

# Evaluar el modelo
resultados_knn = evaluar_modelo(
    "K-Nearest Neighbors (K-NN)",
    modelo_knn,
    X_test,
    y_test
)
```

**Interpretación de resultados:** El modelo K-NN obtuvo un Accuracy del 95,61 %, una Precision del 97,44 %, un Recall del 90,48 % y un F1-score del 93,83 %, lo que refleja un buen desempeño en la clasificación. La matriz de confusión muestra que clasificó correctamente 71 tumores benignos y 38 malignos. Sin embargo, presentó 4 falsos negativos y 1 falso positivo. En general, el modelo logró buenos resultados, aunque dejó de identificar correctamente cuatro tumores malignos.

Naive Bayes:

```python

from sklearn.naive_bayes import GaussianNB

modelo_nb = GaussianNB()

# Entrenar el modelo
modelo_nb.fit(X_train, y_train)

# Evaluar el modelo
resultados_nb = evaluar_modelo(
    "Gaussian Naive Bayes",
    modelo_nb,
    X_test,
    y_test
)
```

**Interpretación de resultados:** El modelo Gaussian Naive Bayes obtuvo un Accuracy del 93,86 %, una Precision del 100 %, un Recall del 83,33 % y un F1-score del 90,91 %. La matriz de confusión muestra que clasificó correctamente 72 tumores benignos y 35 malignos, sin falsos positivos. Sin embargo, presentó 7 falsos negativos, es decir, siete tumores malignos fueron clasificados erróneamente como benignos. Aunque el modelo obtuvo una Precision alta, su capacidad para identificar todos los tumores malignos fue inferior a la de Random Forest y SVM.

### Conclusiones

La comparación de los cinco algoritmos permitió identificar que Random Forest y SVM obtuvieron el mejor desempeño general, con un Accuracy del 97,37 % y un Recall del 92,86 %.
